# Customer Support Ticket Router

A small LangGraph example that classifies a support ticket with keyword rules and routes it to a billing or technical reply node. It uses no LLM or API key.

```python
from typing import Literal, TypedDict

from IPython.display import Image, display
from langgraph.graph import END, START, StateGraph


class TicketState(TypedDict):
    ticket: str
    category: str
    reply: str


def classify_ticket(state: TicketState) -> dict[str, str]:
    """Choose a category using simple, case-insensitive keyword rules."""
    text = state["ticket"].lower()
    billing_keywords = ("bill", "charge", "payment", "refund", "invoice")
    category = "billing" if any(word in text for word in billing_keywords) else "technical"
    return {"category": category}


def route_ticket(state: TicketState) -> Literal["billing", "technical"]:
    return state["category"]  # Selects the matching node name.


def billing_reply(state: TicketState) -> dict[str, str]:
    return {"reply": "I can help with your billing question. Please share the invoice or charge details so we can look into it."}


def technical_reply(state: TicketState) -> dict[str, str]:
    return {"reply": "I can help troubleshoot this technical issue. Please share the steps to reproduce it and any error message."}
```

## Build the graph

The classify node updates the ticket category. A conditional edge sends the state to the matching reply node, and both paths then finish.

```python
builder = StateGraph(TicketState)
builder.add_node("classify", classify_ticket)
builder.add_node("billing", billing_reply)
builder.add_node("technical", technical_reply)

builder.add_edge(START, "classify")
builder.add_conditional_edges("classify", route_ticket)
builder.add_edge("billing", END)
builder.add_edge("technical", END)

ticket_router = builder.compile()
```

## Draw the graph

Render the compiled graph as an image. LangGraph may install its Mermaid rendering dependencies on the first call.

```python
display(Image(ticket_router.get_graph().draw_mermaid_png()))
```

## Try one ticket per route

```python
sample_tickets = [
    "I was charged twice for my monthly subscription.",
    "The app crashes whenever I try to upload a file.",
]

for ticket in sample_tickets:
    result = ticket_router.invoke({"ticket": ticket, "category": "", "reply": ""})
    print(f"Ticket:   {result['ticket']}")
    print(f"Category: {result['category']}")
    print(f"Reply:    {result['reply']}\n")
```
